# Equações Diferenciais (Ordinárias) de Ordem Superior e Sistema de Equações Diferenciais (Ordinárias) de Primeira Ordem

Considere uma equação diferencial de ordem $m$ qualquer

$$y^{(m)}=F(x,y,y^{\prime},y^{\prime\prime},\ldots,y^{(m-1)})$$

Podemos resolver qualquer equação diferencial de ordem elevada reduzindo-a a um sistema de equações diferenciais de primeira ordem. Para tanto,
basta fazermos as seguintes mudanças de variáveis:

$$
\left\{\begin{array} {l}
z_{1}=y\\
z_{1}^{\prime}=y^{\prime}=z_{2}\\
z_{2}^{\prime}=y^{\prime\prime}=z_{3}\\
\vdots\\
z_{m-1}^{\prime}=y^{(m-1)}=z_{m}\\
z_{m}^{\prime}=y^{(m)}=F(x,z_{1},z_{2},z_{3},\ldots,z_{m})
\end{array} \right.
.
$$

## Exemplo 1.

Considere o problema de valor inicial

$$
\left\{\begin{array}{l}
y^{\prime\prime\prime}=F(x,y,y^{\prime},y^{\prime\prime})\\
y(0)=1.1, y^{\prime}(0)=2.2, y^{\prime\prime}(0)=3.3
\end{array} \right.
$$

em que

$$F(x,y,y^{\prime},y^{\prime\prime})=x^{2}+y^{2}\sin(x+y)-(x+1)y^{\prime\prime}-\cos(xy^{\prime})+(x^{2}-1)y.$$

Note que trata-se de uma equação diferencial de ordem 3. Essa pode ser reescrita como o sistema


$$
\left\{\begin{array} {l}
z_{1}=y\\
z_{1}^{\prime}=y^{\prime}=z_{2}\\
z_{2}^{\prime}=y^{\prime\prime}=z_{3}\\
z_{3}^{\prime}=y^{\prime\prime\prime}=F(x,z_{1},z_{2},z_{3})
\end{array} \right.
,
$$


sujeito às condições iniciais $z_{1}(0)=1.1, z_{2}(0)=2.2, z_{3}(0)=3.3$.

# Emprego do Método de Runge-Kutta de Quarta Ordem

Os métodos de Runge-Kutta de ordem $m$ fornecem valores aproximados da solução do problema de valor inicial

$$
\left\{\begin{array}{l}
y^{\prime}=F(x,y)\\
y(x_{0})=y_{0}
\end{array} \right.
,
$$

que coincidem comos valores obtidos através da expansão em série de Taylor de $y$, em torno do ponto $x$, até o termo que inclui $h^{m}$, ou seja,

$$y(x+h)\approx{}y(x)+hf(x,y(x))+\frac{h^{2}}{2!}f^{\prime}(x,y(x))+\cdots+\frac{h^{m}}{m!}f^{(m-1)}(x,y(x)).$$

Nestes métodos o cálculo do valor das derivadas da função $f(x,y(x))$ nos pontos $(x_{n},y_{n})$ será substituído pelo cálculo da função $f(x,y)$ em pontos convenientes, produzindo resultados equivalentes.

O método de Runge-Kutta de ordem $m=1$ recai no método de Euler.

O método de Runge-Kutta de quarta ordem é definido pelas equações

$$y_{n}=y_{n-1}+\frac{h}{6}(K_{1}+2K_{2}+2K_{3}+K_{4}), \qquad n=1,2,\ldots,N,$$

em que

$$K_{1}=F(x_{n-1},y_{n-1}), \qquad K_{2}=F\left(x_{n-1}+\frac{h}{2},y_{n-1}+\frac{h}{2}K_{1}\right),$$

$$K_{3}=F\left(x_{n-1}+\frac{h}{2},y_{n-1}+\frac{h}{2}K_{2}\right), \qquad K_{4}=F(x_{n-1}+h,y_{n-1}+hK_{3}).$$

## Exemplo 2. Movimento não-linear amortecido de um pêndulo

Um pêndulo é modelado por uma massa presa a uma haste rígida de peso desprezível. De acordo com a segunda lei de Newton, quando o pêndulo balança de um lado para o outro, a soma das forças que atuam na massa é igual à massa do pêndulo vezes a aceleração (veja o diagrama de corpo livre na figura).

image.png

Escrevendo a equação de equilíbrio na direção tangencial, obtém-se:

$$-cL\,\frac{d\theta}{dt}-mg\operatorname{sen}\theta=mL\,\frac{d^{2}\theta}{dt^{2}},$$

onde $\theta$ é o ângulo do pêndulo em relação ao eixo vertical, conforme mostrado na figura, $c=0,16\,\mathrm{(N\cdot{}s)/m}$ é o coeficiente de amortecimento, $m=0,5\,\mathrm{kg}$ é a massa, $L=1,2\,\mathrm{m}$ é o comprimento da haste e $g=9,81\,\mathrm{m/s^{2}}$ é a aceleração da gravidade.

Essa equação pode ser reescrita como a seguinte equação diferencial de segunda ordem:

$$\frac{d^{2}\theta}{dt^{2}}=-\frac{c}{m}\frac{d\theta}{dt}-\frac{g}{L}\,\operatorname{sen}\theta.$$

O pêndulo é inicialmente deslocado em $\theta=90^{\circ}$, e então, em $t=0$, ele é solto, tendo velocidade inicial nula ($d\theta/dt=0$). Determine o ângulo do pêndulo em função do tempo, $\theta(t)$, nos primeiros $18\,\mathrm{s}$ após o instante em que ele é solto.

**Observação.** Para valores suficientemente pequenos de $\theta$, podemos utilizar a aproximação linear $\operatorname{sen}\theta\approx\theta$, o que transforma a equação diferencial em uma equação linear, cuja solução pode ser obtida analiticamente. Sem essa aproximação, entretanto, a equação permanece não linear, sendo necessário recorrer a métodos numéricos para obter sua solução.

Então temos em mãos o problema de valor inicial

$$
\left\{\begin{array}{l}
\theta^{\prime\prime}=F(t,\theta,\theta^{\prime})\\
\theta(0)=\pi/2 \quad \mbox{e} \quad \theta^{\prime}(0)=0,
\end{array} \right.
$$

em que

$$F(t,\theta,\theta^{\prime})=-\frac{c}{m}\,\theta^{\prime}-\frac{g}{L}\,\operatorname{sen}\theta.$$

Note que se trata de uma equação diferencial de ordem 2. Essa pode ser reescrita como o sistema

$$
\left\{\begin{array} {l}
z_{1}=\theta\\
z_{1}^{\prime}=\theta^{\prime}=z_{2}\\
z_{2}^{\prime}=\theta^{\prime\prime}=F(t,z_{1},z_{2})
\end{array} \right.
,
$$

sujeito às condições iniciais $z_{1}(0)=\pi/2$ e $z_{2}(0)=0$.

```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------- Parâmetros do sistema ----------------

c = 0.16  # coeficiente de amortecimento, dado em (N.s)/m
m = 0.5   # massa do objeto, dada em kg
L = 1.0   # comprimento do fio, dado em m
g = 9.81  # aceleração da gravidade, dada em m/s^2

def F(t, z):
    z1, z2 = z
    return np.array([z2,
                     - (c / m) * z2 - (g / L) * np.sin(z1)])

# ---------------- Método de Runge-Kutta de 4ª ordem ----------------

def rk4(F, z_0, t0, tf, h):
    t = np.arange(t0, tf+h, h)
    N = len(t) - 1

    z = np.full((N+1, 2), np.nan)

    z[0] = z_0
    for n in range(1, N+1):
        K1 = F(t[n-1], z[n-1])
        K2 = F(t[n-1] + h / 2, z[n-1] + h / 2 * K1)
        K3 = F(t[n-1] + h / 2, z[n-1] + h / 2 * K2)
        K4 = F(t[n-1] + h, z[n-1] + h * K3)
        z[n] = z[n-1] + (h / 6) * (K1 + 2 * K2 + 2 * K3 + K4)
    return t, z

# ---------------- Condições iniciais e passo ----------------

z1_0, z2_0 = np.pi / 2, 0  # ângulo e velocidade angular iniciais
t0, tf = 0, 18             # intervalo de tempo
h = 0.1                    # passo

t, z = rk4(F, np.array([z1_0, z2_0], dtype = float), t0, tf, h)
z1, z2 = z[:, 0], z[:, 1]

# ---------------- Figura ----------------

plt.plot(t, z1, color = "tab:blue", lw = 2)
t_min = min(t) - 0.05*abs(min(t))
t_max = max(t) + 0.05*abs(max(t))
theta_min = min(z1) - 0.05*abs(min(z1))
theta_max = max(z1) + 0.05*abs(max(z1))
plt.axis([t_min, t_max, theta_min, theta_max])  # eixo t entre t_min e t_max; eixo y entre theta_min e theta_max
plt.xlabel("t (s)")        # nomeia o eixo x de "t (s)"
plt.ylabel("theta (rad)")  # nomeia o eixo y de "theta (rad)"
plt.show()
```
