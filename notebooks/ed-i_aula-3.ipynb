# Campos de Direções

```python
import numpy as np
import matplotlib.pyplot as plt


# dados de entrada
def F(x, y):  # a edo é dada por y' = F(x, y)
  F = x + y
  return F


x_min = -2
x_max = 2

y_min = -2
y_max = 2

dx = 0.2
dy = 0.2

comprimento = 0.05


# pontos da malha
x = np.arange(x_min, x_max + dx, dx)
y = np.arange(y_min, y_max + dy, dy)


# construir o campo de direções
plt.figure(figsize=(8, 6))

for xi in x:
  for yi in y:

    # inclinação da solução no ponto (xi, yi)
    inclinacao = F(xi, yi)

    # vetor diretor (1, inclinacao)
    vx = 1
    vy = inclinacao

    # normalização
    norma = np.sqrt(vx**2 + vy**2)

    vx = vx / norma
    vy = vy / norma

    # extremidades do pequeno segmento
    x1 = xi - comprimento * vx
    x2 = xi + comprimento * vx

    y1 = yi - comprimento * vy
    y2 = yi + comprimento * vy

    # desenhar o segmento
    plt.plot([x1, x2], [y1, y2], "k-")


# configurar o gráfico
plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Campo de direções para $y' = F(x,y)$")
plt.grid(True, alpha = 0.3)
plt.show()
```

# Método de Euler (Método de Runge-Kutta de Primeira Ordem)



## Exemplo 3.

```python
import numpy as np
import matplotlib.pyplot as plt


# dados de entrada
def F(x, y): # a edo é dada por y' = F(x, y)
  F = x + y
  return F

x0 = 0 # limite inferior do intervalo
xf = 1 # limite superior do intervalo

h = 0.1 # largura do passo de integração

y0 = 1 # condição inicial y(x0) = y0


# vetor para armazenar x0, x1, ..., xN, em que xn = xn-1 + h para n = 1, 2, ..., N
x = np.arange(x0, xf+h, h)
N = len(x) - 1


# vetor para armazenar y0, y1, ..., yN
y = np.full(N+1, np.nan)


# o método
y[0] = y0
for n in range(1, N+1):
  y[n] = y[n-1] + h * F(x[n-1], y[n-1])


# mostrar os pares (xn, yn) para n = 0, 1, ..., N
print(f"Valor de cada um dos pares (xn, yn) para n = 0, 1, ..., {N}:\n")
for n in range(0, N+1):
  print(f"(x{n}, y{n}) = ({x[n]}, {y[n]})")
print("\n")


# mostrar como um asterisco vermelho no plano cartesiano os pares (xn, yn) para n = 0, 1, ..., N
plt.plot(x, y, "*r")
x_min = min(x) - 0.05*abs(min(x))
x_max = max(x) + 0.05*abs(max(x))
y_min = min(y) - 0.05*abs(min(y))
y_max = max(y) + 0.05*abs(max(y))
plt.axis([x_min, x_max, y_min, y_max]) # eixo x entre x_min e x_max; eixo y entre y_min e y_max
plt.xlabel("x") # nomeia o eixo x de "x"
plt.ylabel("y") # nomeia o eixo y de "y"
plt.show()
```

Comparação com a solução exata, $y(x)=2e^{x}-x-1$:

```python
# mostrar os erros relativos absolutos |yn - y(xn)| / |y(xn)| para n = 0, 1, ..., N
def f(x): # a solução exata do pvi
  y = 2 * np.exp(x) - x - 1
  return y

print(f"Valor de cada um erros relativos absolutos |yn - y(xn)| / |y(xn)| para n = 0, 1, ..., {N}:\n")
for n in range(0, N+1):
  diferenca_relativa = np.abs(y[n] - f(x[n]))/np.abs(f(x[n]))
  print(f"em x = {x[n]}: y{n} - f(x{n}) = {diferenca_relativa}")
```

# Método de Runge-Kutta de Quarta Ordem

Os métodos de Runge-Kutta de ordem $m$ fornecem valores aproximados da solução do problema de valor inicial

$$
\left\{\begin{array}{l}
y^{\prime}=F(x,y)\\
y(x_{0})=y_{0}
\end{array} \right.
,
$$

que coincidem comos valores obtidos através da expansão em série de Taylor de $y$, em torno do ponto $x$, até o termo que inclui $h^{m}$, ou seja,

$$y(x+h)\approx{}y(x)+hf(x,y(x))+\frac{h^{2}}{2!}f^{\prime}(x,y(x))+\cdots+\frac{h^{m}}{m!}f^{(m-1)}(x,y(x)).$$

Nestes métodos o cálculo do valor das derivadas da função $f(x,y(x))$ nos pontos $(x_{n},y_{n})$ será substituído pelo cálculo da função $f(x,y)$ em pontos convenientes, produzindo resultados equivalentes.

O método de Runge-Kutta de ordem $m=1$ recai no método de Euler.

O método de Runge-Kutta de quarta ordem é definido pelas equações

$$y_{n}=y_{n-1}+\frac{h}{6}(P+2Q+2R+S), \qquad l=1,2,\ldots,N,$$

em que

$$P=F(x_{n-1},y_{n-1}), \qquad Q=F\left(x_{n-1}+\frac{h}{2},y_{n-1}+\frac{h}{2}P\right),$$

$$R=F\left(x_{n-1}+\frac{h}{2},y_{n-1}+\frac{h}{2}Q\right), \qquad S=F(x_{n-1}+h,y_{n-1}+hR).$$

```python
import numpy as np
import matplotlib.pyplot as plt


# dados de entrada
def F(x, y): # a edo é dada por y' = F(x, y)
  F = x + y
  return F

x0 = 0 # limite inferior do intervalo
xf = 1 # limite superior do intervalo

h = 0.1 # largura do passo de integração

y0 = 1 # condição inicial y(x0) = y0


# vetor para armazenar x0, x1, ..., xN, em que xn = xn-1 + h para n = 1, 2, ..., N
x = np.arange(x0, xf+h, h)
N = len(x) - 1


# vetor para armazenar y0, y1, ..., yN
y = np.full(N+1, np.nan)


# o método
y[0] = y0
for n in range(1, N+1):
  P = F(x[n-1], y[n-1])
  Q = F(x[n-1] + h/2, y[n-1] + (h/2) * P)
  R = F(x[n-1] + h/2, y[n-1] + (h/2) * Q)
  S = F(x[n-1] + h, y[n-1] + h * R)
  y[n] = y[n-1] + (h/6) * (P + 2 * Q + 2 * R + S)


# mostrar os pares (xn, yn) para n = 0, 1, ..., N
print(f"Valor de cada um dos pares (xn, yn) para n = 0, 1, ..., {N}:\n")
for n in range(0, N+1):
  print(f"(x{n}, y{n}) = ({x[n]}, {y[n]})")
print("\n")


# mostrar como um asterisco vermelho no plano cartesiano os pares (xn, yn) para n = 0, 1, ..., N
plt.plot(x, y, "*r")
x_min = min(x) - 0.05*abs(min(x))
x_max = max(x) + 0.05*abs(max(x))
y_min = min(y) - 0.05*abs(min(y))
y_max = max(y) + 0.05*abs(max(y))
plt.axis([x_min, x_max, y_min, y_max]) # eixo x entre x_min e x_max; eixo y entre y_min e y_max
plt.xlabel("x") # nomeia o eixo x de "x"
plt.ylabel("y") # nomeia o eixo y de "y"
plt.show()
```

Comparação com a solução exata, $y(x)=2e^{x}-x-1$:

```python
# mostrar os erros relativos absolutos |yn - y(xn)| / |y(xn)| para n = 0, 1, ..., N
def f(x): # a solução exata do pvi
  y = 2 * np.exp(x) - x - 1
  return y

print(f"Valor de cada um erros relativos absolutos |yn - y(xn)| / |y(xn)| para n = 0, 1, ..., {N}:\n")
for n in range(0, N+1):
  diferenca_relativa = np.abs(y[n] - f(x[n]))/np.abs(f(x[n]))
  print(f"em x = {x[n]}: y{n} - f(x{n}) = {diferenca_relativa}")
```
